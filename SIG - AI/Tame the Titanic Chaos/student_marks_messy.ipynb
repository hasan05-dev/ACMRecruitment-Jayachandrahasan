```python
import pandas as pd

# load dataset
students = pd.read_csv("student_data_part1.csv")

# inspect dataset
print(students.info())
print(students.describe(include='all'))
print(students.isnull().sum())
print(students.shape)
```

Used **.info()** for column types, **.describe()** for stats, **.isnull().sum()** for missing values and **.shape** for dimensions.

written **include='all'** in **.describe()** to get an overview of both numeric and alphabetical data.

```python
# Standardize gender values
students['Gender'] = students['Gender'].astype(str).str.strip().str.lower()
students['Gender'] = students['Gender'].replace({'f':'female','m':'male','female':'female','male':'male','girl':'female','boy':'male','0':'unknown','1':'unknown'})

# Correct data type
for col in ['Math','Science','English','Total']:
    students[col] = students[col].astype(str).str.replace("marks","",regex=False).str.strip()
    students[col] = pd.to_numeric(students[col], errors='coerce')

# Correct data type
students = students.drop_duplicates()

print(students.isnull().sum())
```

astype(str) - converts all values into strings to perform string operations

.str.strip() - removes extra spaces

.str.lower() - converts everything to lowercase

.replace({...}) - maps messy values into consistent categories

.str.replace("marks","") - removes the word "marks" from entries

pd.to_numeric(..., errors='coerce') - converts cleaned strings into numbers


the dataset has no missing values after the cleaning step in 1c. hence, step 1d (The Missing Pieces) us not needed.
